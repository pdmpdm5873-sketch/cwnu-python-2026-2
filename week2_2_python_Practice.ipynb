````python
import ColabTurtle.Turtle as t
t.initializeTurtle()
t.speed(10)

# 각도 리스트
angles = [100, 120, 144, 150, 160]
colors = ["red", "orange", "yellow", "green", "blue"]

# 각 각도별로 도형 그리기
for i in range(len(angles)):
    t.penup()
    t.goto(50 + (i * 150), 200) # 겹치지 않게 위치 이동
    t.pendown()
    t.color(colors[i])

    angle = angles[i]
    # 넉넉하게 20번 반복하여 도형이 어떻게 변하는지 관찰
    for _ in range(20):
        t.forward(60)
        t.right(angle)

print("테스트 완료: 각도별로 다른 모양이 생성되었습니다.")
````

````python
import ColabTurtle.Turtle as t
t.initializeTurtle()
t.speed(10)

# 그릴 별들의 각도 리스트
star_angles = [144, 150, 160, 170, 175]

# 바깥쪽 for문: 각도와 위치를 바꿉니다.
for i in range(len(star_angles)):
    t.penup()
    t.goto(100 + (i * 120), 250) # 각 별의 시작 위치 조절
    t.pendown()

    current_angle = star_angles[i]

    # 안쪽 for문: 실제로 하나의 별(도형)을 그립니다.
    for j in range(20): # 넉넉히 반복해서 닫히게 함
        t.forward(50)
        t.right(current_angle)

print("중첩 for문을 이용해 5개의 별 패턴을 그렸습니다!")
````

이런 식으로 `for`문 안에 `for`문을 넣으면, 바깥쪽 루프가 한 번 실행될 때마다 안쪽 루프가 전체 횟수만큼 반복됩니다.

### 1. 각도별 결과 정리
- **100도**: 도형이 닫히지 않고 계속 엇갈리며 복잡한 패턴을 만듭니다.
- **120도**: 세 번의 선을 그리면 정확히 **정삼각형**이 되어 닫힙니다 (뾰족한 점 3개).
- **144도**: 다섯 번의 선을 그리면 우리가 아는 **별(오각형 별)** 모양이 되어 닫힙니다 (뾰족한 점 5개).
- **150도**: 12번을 반복해야 닫히는 톱니바퀴 같은 별 모양이 됩니다.
- **160도**: 18번을 반복해야 닫히는 더 촘촘한 별 모양이 됩니다.

### 2. "닫히는 조합"의 규칙
도형이 처음 시작점으로 돌아와 닫히기 위해서는 **(회전한 각도 × 반복 횟수)가 360의 배수(360, 720, 1080...)**가 되어야 합니다.

- **삼각형**: 120도 × 3번 = 360도
- **오각형 별**: 144도 × 5번 = 720도 (360 × 2)
- **정사각형**: 90도 × 4번 = 360도

따라서 `range(n)`의 `n`값이 `(360 * 바퀴수 / 각도)`를 만족하는 최소 정수가 될 때 도형이 예쁘게 닫히게 됩니다!
